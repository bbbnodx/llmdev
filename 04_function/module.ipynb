## 標準ライブラリの活用
- math
- datetime
- random
- os

```python
import math

# 平方根の計算
print(math.sqrt(25))

# 指数関数の計算
print(math.exp(1))

# 定数の使用
print(math.pi)
```

```python
import datetime

# 現在の日付と時刻を取得
now = datetime.datetime.now()
print(now)

# 特定の日付を作成
birthday = datetime.datetime(1995, 5, 17)
print(birthday)

# 日付の差を計算
date1 = datetime.datetime(2023, 1, 1)
date2 = datetime.datetime(2024, 1, 1)
diff = date2 - date1
print(diff.days)
```

```python
import random

# ランダムな整数を生成
print(random.randint(1, 10))

# リストをシャッフル
my_list = [1, 2, 3, 4, 5]
random.shuffle(my_list)
print(my_list)

# リストからランダムに要素を選択
print(random.choice(my_list))
```

```python
import os

# カレントディレクトリの取得
print(os.getcwd())

# 新しいディレクトリを作成
os.mkdir("new_folder")

# ファイルの存在確認
print(os.path.exists("example.txt"))
```

```python
import urllib.request

# URLを指定してデータを取得
url = "https://techacademy.jp"
response = urllib.request.urlopen(url)

# HTMLデータを読み込み、UTF-8文字列にデコード
html = response.read().decode("utf-8")
print(html[:500])
```
