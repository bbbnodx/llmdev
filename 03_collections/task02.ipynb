## 課題 単語の文字数をカウントしよう
### 仕様
英単語の入力を受け付けます。未入力でリターン（エンター）キーを押されるまで、英単語の受付を繰り返します。受付が完了したら、入力された英単語をABC順に並べ替えて表示します。また、すべての英単語に含まれるアルファベットごとの数も表示します。

```python
vocab_list = []
alpha_count_dict = dict.fromkeys("abcdefghijklmnopqrstuvwxyz", 0)

# 英単語の入力ループ
while True:
    vocab = input("英単語を入力してください")
    if len(vocab) == 0:
        break
    if not vocab.isalpha():
        print("アルファベットで入力してください")
        continue
    vocab = vocab.lower()
    vocab_list.append(vocab)
    for char in vocab:
        alpha_count_dict[char] += 1

vocab_list.sort()

# 出力
print("入力した英単語:", vocab_list)
for char, count in alpha_count_dict.items():
    if count > 0:
        print(f"{char}が{count}個ありました")
```
