## 課題 学生の成績分析ツールを作成しよう
### 仕様
- 学生の成績をリストとして定義します。以下のソースコードを使用してください。  
  `grades = [85, 90, 78, 92, 88]`
- 成績のリストを使用して以下を行ってください。
- 成績の平均点を計算します。
- 成績の最高点と最低点を求めます。
- 成績が平均以上の学生数をカウントします。
- 結果を出力してください。

```python
grades = [85, 90, 78, 92, 88]

grades_average = sum(grades) / len(grades)
grades_max = max(grades)
grades_min = min(grades)
count_above_average = len([elm for elm in grades if elm >= grades_average])

print(f"平均点: {grades_average:.1f}")
print(f"最高点: {grades_max}")
print(f"最低点: {grades_min}")
print(f"平均以上の学生数: {count_above_average}")
```
