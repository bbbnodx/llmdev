```python
s = "abcdefghijklmnopqrstuvwxyz"

print(s[0])  # => a
print(s[1])  # => b
print(s[25])  # => z

print(s[-1])  # => z
print(s[-3])  # => x
print(s[-26])  # => a
```

```python
# 以下のコードはIndexErrorとなる
print(s[26])
print(s[-27])
```

```python
# スライス
s = "abcdefghijklmnopqrstuvwxyz"

print(s[0:7])
print(s[21:26])

print(s[-5:-1])
```

```python
# スライスにおける開始または終了位置の省略
s = "abcdefghijklmnopqrstuvwxyz"

print(s[:10])
print(s[10:])
print(s[:])
```

```python
# ステップの指定
s = "abcdefghijklmnopqrstuvwxyz"

print(s[0:10:2])
print(s[1:20:3])
print(s[-1:-10:-2])
```

```python
# 回文の判定
s = "にんてんどううどんてんに"
if s == s[::-1]:
    print(f"「{s}」は回文です！")
else:
    print(f"「{s}」は回文ではないです。")

s = "Pythonは便利だが遅い"
if s == s[::-1]:
    print(f"「{s}」は回文です！")
else:
    print(f"「{s}」は回文ではないです。")

s = "no devil lived on"
if s == s[::-1]:
    print(f"「{s}」は回文です！")
else:
    print(f"「{s}」は回文ではないです。")
```

```python
# シーケンスの反復処理
s = "TechAcademy"
for char in s:
    print(char)
```

```python
s = "TechAcademy"
for char in s:
    print(char.upper())
```

```python
s = "TechAcademy"
vowels = "aeiouAEIOU"  # 母音のリスト
for char in s:
    if char in vowels:
        print(char)
```

```python
# 長さの取得
s = "TechAcademy"
length = len(s)
print(f"文字列の長さ: {length}")
```

```python
s = "Hello, world!"
if len(s) > 10:
    print("10文字を超える長い文字列です。")
else:
    print("10文字以内の短い文字列です。")
```

```python
s = ""
if len(s) == 0:
    print("文字列は空です。")
else:
    print("文字列は空ではありません。")
```
