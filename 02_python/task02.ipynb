## 課題 料込価格を計算しよう
### 仕様
入力された税抜価格をもとに、消費税込の価格と送料を合計した「送料込価格」を計算し、結果を表示するソースコードを作成してください。

- 税込価格が2000円以上の場合は送料無料、2000円未満の場合は350円とします。  
- 消費税率は10%です。  
- 税抜価格の入力には input() 関数を使用してください。  
- 結果の表示にはf文字列を使用してください。

```python
price = int(input("税抜価格(円)を入力してください: "))
price_w_tax = int(price * 1.1) # 税込価格

if price_w_tax >= 2000:
    print("送料は無料です")
    price_w_shipping = price_w_tax
else:
    print("送料は350円です")
    price_w_shipping = price_w_tax + 350

print(f"送料込税込価格は{price_w_shipping}円です。")
```
