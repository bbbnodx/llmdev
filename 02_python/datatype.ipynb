```python
# 整数の代入
a = 10
b = -5
c = 0

# 整数同士の計算
sum_result = a + b
product_result = a * b

print("a:", a)
print("b:", b)
print("c:", c)
print("合計:", sum_result)
print("積:", product_result)
```

```python
# 除算の例
division_result = a / 3
integer_division_result = a // 3

print("除算結果:", division_result)
print("整数除算結果:", integer_division_result)
```

## 書式指定文字列

```python
# 小数点以下の桁数
pi = 3.14159
print(f"小数点以下の桁数: {pi:.2f}")

# カンマ区切り
large_number = 1234567890
print(f"カンマ区切り: {large_number:,}")

# 文字列の出力幅
name = "Alice"
print(f"文字列の幅「{name:10}」")

# ゼロ埋め
num = 42
print(f"ゼロ埋め: {num:05}")

# パーセント表示
accuracy = 0.856
print(f"パーセント表示: {accuracy:.1%}")

# 揃え
name = "Alice"
print(f"左揃え　「{name:<10}」")
print(f"右揃え　「{name:>10}」")
print(f"中央揃え「{name:^10}」")

# 日時のフォーマット
from datetime import datetime
now = datetime.now()
print(f"現在日時: {now:%Y-%m-%d %H:%M:%S}")
```
